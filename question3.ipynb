```python
from qiskit import QuantumCircuit
import numpy as np
from qiskit.quantum_info import SparsePauliOp
from qiskit.quantum_info import Statevector
from scipy.optimize import minimize
```

```python
def ansatz(t1=0, t2=0):
    qc = QuantumCircuit(2)
    qc.ry(t1, 0)
    qc.ry(t2, 1)
    return qc
```

```python
def ansatz_cnot(t1=0,t2=0):
    qc = QuantumCircuit(2)
    qc.ry(t1,0)
    qc.ry(t2,1)
    qc.cx(0,1)
    return qc
```

```python
pauli_operators = ["II", "IZ", "ZI","ZZ", "XX"] 
coeffs = [-1.0523732458, 0.3979374248, -0.3979374248, -0.0112801043, 0.1809311998]
```

```python
H = SparsePauliOp(pauli_operators,coeffs)
```

```python
def cost(thethas,cnot):
    t1,t2 = thethas
    if cnot:
        circuit = ansatz_cnot(t1,t2)
    else :
        circuit = ansatz(t1,t2)
    state = Statevector.from_circuit(circuit)
    energy = np.real(state.expectation_value(H))
    return energy


E_0 = cost([0,0],True)
print(f"energy with (0,0) : {E_0}")
```

```python
optimise = minimize(cost,[0.0,0.0],args=(True),method='COBYLA')
theta1 , theta2 = optimise.x
E_vqe = optimise.fun
```

```python
E_0_exact = -1.85727503
Delta_E = E_vqe - E_0_exact
print(Delta_E)
```
